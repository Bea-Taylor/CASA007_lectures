## Figures for CASA007 Lecture 4

This nb generates the figures for Lecture 3 - Linear Algebra. All figs generated are saved to the folder 'L4_images'. 

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from matplotlib.patches import FancyArrowPatch, Circle, Rectangle
import numpy as np
import textwrap
```

```python
# generate a plot of a linear line and annotate it with the equation of the line 
x = np.linspace(-4, 10, 100)
y = 2*x + 3
plt.figure(figsize=(8,6))
plt.plot(x, y, color="#e16fca")
plt.xlabel(r"$x$")
plt.ylabel(r"$y$")
plt.xlim(-4, 10)
plt.text(2.1, 13, r"$y=2x+3$", fontsize=14, fontweight="bold", color="#8EA358")
plt.axhline(0, color='black',linewidth=1, ls='--')
plt.axvline(0, color='black',linewidth=1, ls='--')
plt.grid(color = 'gray', linestyle = '--', linewidth = 0.5)
plt.legend()
plt.savefig('L4_images/linear_function.png', dpi=300, bbox_inches='tight', transparent=True)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(6, 6))

n_rows, n_cols = 4, 4
for i in range(n_rows):
    for j in range(n_cols):
        ax.add_patch(Rectangle((j, i), 1, 1, fill=False))
        ax.text(j + 0.5, i + 0.5, rf"$a_{{{i+1},{j+1}}}$",
                ha="center", va="center", fontsize=12)

# highlight a_{3,2}
ax.add_patch(Rectangle((1, 2), 1, 1, fill=False, linewidth=3, linestyle="--"))

# column labels and arrow
for j in range(n_cols):
    ax.text(j + 0.5, -0.25, f"{j+1}", ha="center", va="center", fontsize=12)
ax.text(2, -0.6, r"columns $n$ (across)", ha="center", va="center", fontsize=12)
ax.annotate("", xy=(n_cols, -0.5), xytext=(0, -0.5), arrowprops=dict(arrowstyle="->"))

# row labels and arrow
for i in range(n_rows):
    ax.text(-0.25, i + 0.5, f"{i+1}", ha="center", va="center", fontsize=12, rotation=90)
ax.text(-0.6, 2, r"rows $m$ (down)", ha="center", va="center", rotation=90, fontsize=12)
ax.annotate("", xy=(-0.5, n_rows), xytext=(-0.5, 0), arrowprops=dict(arrowstyle="->"))

ax.set_xlim(-1, n_cols + 0.5)
ax.set_ylim(-1, n_rows + 0.5)
ax.invert_yaxis()
# ax.set_aspect('equal', adjustable='box')
ax.set_xticks([]); ax.set_yticks([])
# ax.set_title("Along the corridor, down the stairs", pad=20)

for spine in ax.spines.values():
    spine.set_visible(False)

plt.savefig('L4_images/matrix_elements.png', dpi=300, bbox_inches='tight', transparent=True)
plt.show()
```
